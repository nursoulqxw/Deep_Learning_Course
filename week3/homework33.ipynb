## Task 1

```python
import numpy as np
X = np.array([[1., 0.], [0., 1.], [1., 1.]])
y = np.array([[1.], [2.], [1.]])
W1 = np.array([[1.,
-1.], [-1., 1.]])
b1 = np.array([0.5, 0.5])
W2 = np.array([[1.], [2.]])
b2 = np.array([0.])


```

```python
z1 = X @ W1 + b1
H = np.maximum(0, z1)
yhat = H @ W2 + b2

print("z1 = ", z1)
print("h = ", H)
print("YHAT = ", yhat)


residuals = yhat - y

s_r = residuals**2

mse = s_r.mean()

print("residuals     :", residuals)
print("sqared errors :", s_r)
print("mse           :", mse)

assert yhat.shape == y.shape
```

```python
#without ReLU
z2 = X @ W1 + b1
yhat_lin = z2 @ W2 + b2
residuals_lin = yhat_lin - y
s_r_lin = residuals_lin**2
mse_lin = (s_r_lin).mean()
print("without relu     :", yhat_lin)
print("mse without relu :", mse_lin)
```

```python
W_eff = W1 @ W2
b_eff = b1 @ W2 + b2

yhat_eff = W_eff + b_eff

print("yhat_eff :", yhat_eff)
```

The first group touches X, and the second group doesn't depend on X at all. That's exactly the form of one linear layer: $XW + b$. The two layers merged into one.

## Task 2

```python
y2 = np.array([1,2,4])
Yhat2 = np.array([0,2,10])

residuals2 = Yhat2 - y2

squared_error2 = residuals2**2

mse = np.mean(squared_error2)


mae = np.mean(np.abs(residuals2))

print("mse :", mse)
print("mae :", mae)
```

MSE is much more sensitive to the large error than MAE. When the third prediction's error doubled from 3 to 6, MAE rose from 1.33 to 2.33 (about 1.75×), while MSE rose from 3.33 to 12.33 (about 3.7×). MAE grows linearly with the error, so the outlier adds proportionally. MSE squares the error, so doubling it made that term 4× larger (9 became 36), and one sample ended up dominating the whole loss.

```python
yi = np.array([1,0])
y_A = np.array([0.8,0.2])
y_B = np.array([0.6, 0.4])

def bce(yi, p):
    L = -np.mean(yi * np.log(p) + (1-yi) * np.log(1-p))
    return L

bce_A = bce(yi, y_A)
bce_B = bce(yi, y_B)

pred_A = (y_A > 0.5).astype(int)
pred_B = (y_B > 0.5).astype(int)

acc_A = np.mean(pred_A == yi)
acc_B = np.mean(pred_B == yi)

print("bce_A :", bce_A, "; pred_A :", pred_A, "; acc_A :", acc_A)
print("bce_B :", bce_B, "; pred_B :", pred_B, "; acc_B :", acc_B)

```
