```python
import numpy as np

y = np.array([0.0, 0.0, 0.0, 1.0])
y_pred = np.array([0.1, 0.2, 0.6, 0.1])


def cross_entropy(y, y_pred):

    loss = - (y* np.log(y_pred) + (1-y) * np.log(1-y_pred))

    return np.sort(loss), np.mean(loss)


sorted_loss, meann_loss = cross_entropy(y, y_pred)

print("losses sorted", sorted_loss)
print("meann sorted", meann_loss)



```
