```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
X = torch.tensor([[1., 2., 0., 1.],
                  [0., 1., 3., 2.],
                  [2., 2., 1., 0.],
                  [1., 0., 2., 3.]])

K = torch.tensor([[1., 0., -1.],
                  [1., 0., -1.],
                  [1., 0., -1.]])

# Your code here

Y = F.conv2d(
    X.view(1,1,4,4),
    K.view(1,1,3,3)
)

print(Y)
```

```python
# Your code here

layer = nn.Conv2d(1,8,3, padding = 0, stride = 1)
x = torch.randn(1,1,28,28)
y = layer(x)


print("a:", y.shape)
print("prs:", sum(p.numel() for p in layer.parameters()))


layer = nn.Conv2d(1,8, 3, padding=1, stride=1)
x = torch.randn(1, 1, 28, 28)
y = layer(x)

print("a:", y.shape)
print("prs:", sum(p.numel() for p in layer.parameters()))


layer = nn.Conv2d(1, 8, 3, padding=1, stride=2)
x = torch.randn(1, 1, 28, 28)
y = layer(x)

print("c:", y.shape)
print("prs:", sum(p.numel() for p in layer.parameters()))


layer = nn.Conv2d(3, 16, 5, padding=2, stride=1)
x = torch.randn(1, 3, 32, 32)
y = layer(x)

print("d:", y.shape)
print("prs:", sum(p.numel() for p in layer.parameters()))
```

```python


model = nn.Sequential(
    nn.Conv2d(1, 8, kernel_size=3, padding=1),
    nn.ReLU(),
    nn.MaxPool2d(2),
    nn.Conv2d(8, 16, kernel_size=3, padding=1),
    nn.ReLU(),
    nn.MaxPool2d(2),
    nn.Flatten(),
    nn.Linear(16 * 7 * 7, 10)
)


x = torch.randn(4, 1, 28, 28)

for layer in model:
    x = layer(x)
    print(layer, "->", x.shape)

total = 0

for layer in model:
    params = sum(p.numel() for p in layer.parameters())
    print(layer, "parameters:", params)
    total += params

print("Total parameters:", total)

print("MLP parameters:", 101770)
print("CNN parameters:", total)
```
